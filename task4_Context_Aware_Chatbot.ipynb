```python

with open("kb.txt", "r") as f:
    kb_lines = f.readlines()

kb_lines = [line.strip() for line in kb_lines]
kb_lines
```

```python
def simple_chatbot(user_input):
    user_input_lower = user_input.lower()
    matches = []

    for line in kb_lines:
        if any(word in user_input_lower for word in line.lower().split()):
            matches.append(line)

    if matches:
        return "\n".join(matches[:3])  # top 3 matches
    else:
        return "Sorry, I do not have an answer for that."
```

```python
user_questions = [
    "How can I reset my password?",
    "I want a refund for my order.",
    "App keeps crashing, what should I do?",
    "Are new features added?",
    "Where do I check my invoice?"
]

for q in user_questions:
    print(f"Q: {q}")
    print(f"A: {simple_chatbot(q)}\n")
```
